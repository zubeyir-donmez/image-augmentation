```python
import cv2
import numpy as np

# Create a black image (50x50 pixels)
img = np.zeros((50, 50, 3), dtype=np.uint8)

# Draw a white rectangle on the image
cv2.rectangle(img, (10, 10), (40, 40), (255, 255, 255), -1)

# Print a message to confirm cv2 is working
print("cv2 imported and used successfully!")
```
